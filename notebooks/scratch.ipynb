```python
import pandas as pd
f = pd.read_json("results/forecast.json")
a = pd.read_parquet("data/processed/zone_day_demand.parquet")
print("forecast rows:", len(f), "| cols:", list(f.columns))
print("actual cols:", list(a.columns))
print(f.head(3)); print(f.dtypes)
f["date"] = pd.to_datetime(f["date"]); a["date"] = pd.to_datetime(a["date"])
print("forecast distinct dates:", f["date"].nunique(), f["date"].min(), f["date"].max())
f["PULocationID"] = f["PULocationID"].astype(int); a["PULocationID"] = a["PULocationID"].astype(int)
m = f.merge(a, on=["PULocationID", "date"], suffixes=("_pred", "_actual"))
print("merged rows:", len(m), "| cols:", list(m.columns))
```
